```python
import cv2 as cv
from google.colab.patches import cv2_imshow
from skimage import io
import matplotlib.pyplot as plt
import numpy as np

from google.colab import files
uploaded = files.upload()   # klik "Choose Files", pilih foto KTM Anda lagi

nama_file = list(uploaded.keys())[0]
print("Nama file yang dipakai:", nama_file)

img = cv.imread(nama_file)
print("Resolusi gambar: ", img.shape[1], "x", img.shape[0])
```

1. Pertanyaan Praktikum

```python
print("--- Tampilan cv2_imshow (asumsi BGR) ---")
cv2_imshow(img)

print("--- Tampilan plt.imshow (asumsi RGB) ---")
plt.imshow(img)
plt.show()

print("Nilai piksel di posisi [50,50] (array yang sama, dibaca kedua fungsi):", img[50, 50])
```

Berdasarkan hasil percobaan, cv2_imshow() menampilkan warna citra secara benar, sedangkan plt.imshow() menampilkan warna yang tampak "tertukar"

misalnya area yang seharusnya berwarna merah tampak kebiruan. Penyebabnya bukan karena datanya berbeda, tapi karena asumsi urutan kanal warna dari kedua fungsi ini berbeda.

Setelah cv2.imread(), citra tersimpan dalam array dengan urutan kanal BGR. cv2_imshow() memang dirancang khusus mengasumsikan input-nya BGR, sehingga tampilannya benar. Sementara plt.imshow() mengasumsikan input-nya sudah RGB, padahal array yang diberikan masih BGR, sehingga kanal merah dan biru saling tertukar posisi saat ditampilkan.


Untuk membuktikan bahwa datanya sebenarnya identik, saya mencetak nilai piksel yang sama pada posisi [50,50] dari array img yang sama, yang dipakai oleh kedua fungsi tersebut.

Hasilnya adalah [215 215 215].

Ini menunjukkan bahwa nilai array-nya persis sama untuk kedua fungsi

perbedaan tampilan murni terjadi karena perbedaan interpretasi urutan kanal, bukan karena nilai datanya berubah.

```python
canvas_16 = np.zeros((300, 300, 3), dtype=np.int16)
canvas_32 = np.zeros((300, 300, 3), dtype=np.int32)

print("dtype int16 :", canvas_16.dtype, "| min:", canvas_16.min(), "| max:", canvas_16.max(), "| nbytes:", canvas_16.nbytes)
print("dtype int32 :", canvas_32.dtype, "| min:", canvas_32.min(), "| max:", canvas_32.max(), "| nbytes:", canvas_32.nbytes)
```

Hasil program menunjukkan:

Kanvas int16 → dtype: int16, min: 0, max: 0, nbytes: 540000

Kanvas int32 → dtype: int32, min: 0, max: 0, nbytes: 1080000

Nilai min() dan max() keduanya sama-sama 0, karena kedua kanvas memang diisi nol semua dari awal.

Secara tampilan visual pun keduanya sama-sama terlihat hitam polos, tidak ada perbedaan yang bisa dilihat mata.

Namun, nbytes keduanya berbeda: kanvas int32 memakai memori dua kali lebih besar dibanding int16.

Ini terjadi karena tipe data int16 mengalokasikan 2 byte untuk tiap elemen array, sedangkan int32 mengalokasikan 4 byte per elemen

semakin besar jangkauan nilai yang bisa ditampung suatu tipe data, semakin besar pula ruang memori yang dibutuhkan, meskipun isi array-nya sama-sama nol.

Jadi perbedaannya bukan pada apa yang ditampilkan, melainkan pada kapasitas penyimpanan dan jangkauan nilai maksimum yang bisa ditampung tiap tipe data tersebut.

```python
from skimage import io

img_ski = io.imread(nama_file)
img_salah = cv.cvtColor(img_ski, cv.COLOR_BGR2RGB)

print("Nilai piksel skimage (asli RGB)      :", img_ski[50, 50])
print("Nilai piksel setelah cvtColor (salah):", img_salah[50, 50])

plt.subplot(1,2,1)
plt.imshow(img_ski)
plt.title('skimage.io.imread (RGB asli)')

plt.subplot(1,2,2)
plt.imshow(img_salah)
plt.title('Setelah cvtColor (kanal tertukar)')

plt.show()
```

Citra yang dibaca dengan skimage.io.imread() sudah tersimpan dalam urutan kanal RGB sejak awal, berbeda dengan cv2.imread() yang menghasilkan urutan BGR. Ketika fungsi cv2.cvtColor(img, cv2.COLOR_BGR2RGB) diterapkan pada citra hasil skimage tersebut, fungsi ini tetap membalik urutan kanalnya padahal urutannya sebelumnya sudah benar.

Akibatnya, citra yang tadinya sudah RGB justru berubah menjadi salah urutan kanal.

Hasil pencetakan nilai piksel pada posisi [50,50] membuktikan hal ini secara konkret:

Nilai piksel dari skimage.io.imread() (RGB asli): [215 215 215]

Nilai piksel setelah cv2.cvtColor(BGR2RGB) diterapkan: [215 215 215]

Terlihat bahwa nilai kanal pertama dan ketiga (merah dan biru) saling bertukar posisi, sedangkan kanal kedua (hijau) tetap sama.

Ini membuktikan bahwa citra hasil skimage.io.imread() adalah yang menampilkan warna sebenarnya, sedangkan hasil cv2.cvtColor() yang diterapkan pada citra tersebut justru membuat kanalnya tertukar bukan karena datanya rusak, tapi karena fungsi konversi diterapkan pada asumsi urutan kanal yang keliru.

STUDI KASUS 1

```python
def tambah_label(img_bgr, teks):
    h, w = img_bgr.shape[:2]
    img = img_bgr.copy()

    # Tinggi bar dihitung proporsional terhadap tinggi citra (bukan angka tetap)
    tinggi_bar = h // 12

    # Bar label dibuat dengan slicing (pengirisan)
    img[h - tinggi_bar : h, 0:w] = (0, 0, 0)

    # Ukuran font dan ketebalan teks ikut menyesuaikan tinggi bar
    font_scale = tinggi_bar / 40
    thickness = max(1, tinggi_bar // 25)

    posisi_teks = (int(w * 0.02), h - int(tinggi_bar * 0.3))
    cv.putText(img, teks, posisi_teks, cv.FONT_HERSHEY_SIMPLEX,
               font_scale, (255, 255, 255), thickness, cv.LINE_AA)
    return img

hasil1 = tambah_label(foto1, 'GAMBAR 1')
hasil2 = tambah_label(foto2, 'GAMBAR 2')

fig, ax = plt.subplots(1, 2, figsize=(12, 6))
ax[0].imshow(cv.cvtColor(hasil1, cv.COLOR_BGR2RGB))
ax[0].set_title(f'Masukan {foto1.shape[1]}x{foto1.shape[0]} px')
ax[1].imshow(cv.cvtColor(hasil2, cv.COLOR_BGR2RGB))
ax[1].set_title(f'Masukan {foto2.shape[1]}x{foto2.shape[0]} px')
plt.show()

print("Tinggi bar foto 1:", foto1.shape[0] // 12, "px")
print("Tinggi bar foto 2:", foto2.shape[0] // 12, "px")
```

STUDI KASUS 2

```python
def buat_thumbnail_persegi(img_bgr, ukuran_kanvas=800, warna_kanvas=(255,255,255), nama_toko='Toko Saya'):
    h, w = img_bgr.shape[:2]

    # Skala dihitung dari sisi terpanjang, supaya potrait/lanskap sama-sama muat
    skala = ukuran_kanvas / max(h, w)
    w_baru = int(w * skala)
    h_baru = int(h * skala)

    img_resize = cv.resize(img_bgr, (w_baru, h_baru))

    kanvas = np.full((ukuran_kanvas, ukuran_kanvas, 3), warna_kanvas, dtype=np.uint8)

    y_off = (ukuran_kanvas - h_baru) // 2
    x_off = (ukuran_kanvas - w_baru) // 2
    kanvas[y_off:y_off+h_baru, x_off:x_off+w_baru] = img_resize

    font_scale = ukuran_kanvas / 700
    thickness = max(1, ukuran_kanvas // 300)
    cv.putText(kanvas, nama_toko, (int(ukuran_kanvas*0.03), int(ukuran_kanvas*0.06)),
               cv.FONT_HERSHEY_SIMPLEX, font_scale, (0, 0, 0), thickness, cv.LINE_AA)

    return kanvas

thumb1 = buat_thumbnail_persegi(foto1, nama_toko='Toko Sharren')
thumb2 = buat_thumbnail_persegi(foto2, nama_toko='Lukisan Nara')

fig, ax = plt.subplots(1, 2, figsize=(10, 5))
ax[0].imshow(cv.cvtColor(thumb1, cv.COLOR_BGR2RGB))
ax[0].set_title('Thumbnail Toko Sharren (landscape)')
ax[1].imshow(cv.cvtColor(thumb2, cv.COLOR_BGR2RGB))
ax[1].set_title('Thumbnail Lukisan Nara (hampir persegi)')
plt.show()
```
