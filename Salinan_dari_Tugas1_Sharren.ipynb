```python
# ==========================================
# STEP 1: Import Library
# ==========================================
import cv2
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from PIL import Image
import skimage
from urllib.request import Request, urlopen

# ==========================================
# STEP 2: Read & Display Images from URLs
# ==========================================
urls = [
    "https://images.unsplash.com/photo-1543466835-00a7907e9de1?w=800",
    "https://images.unsplash.com/photo-1514888286974-6c03e2ca1dba?w=800",
    "https://images.unsplash.com/photo-1533738363-b7f9aef128ce?w=800",
]

headers = {"User-Agent": "Mozilla/5.0"}
images = []
original_rgb_images = []

for url in urls:
    # Read image from URL
    req_stmt = Request(url, headers=headers)
    req = urlopen(req_stmt)
    arr = np.asarray(bytearray(req.read()), dtype=np.uint8)
    image = cv2.imdecode(arr, cv2.IMREAD_COLOR)

    # Resize image to half size
    image = cv2.resize(image, (0, 0), fx=0.5, fy=0.5)

    # Convert color to RGB
    image_2 = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    original_rgb_images.append(image_2.copy())

    # Concatenate image secara horizontal (BGR | RGB)
    final_frame = cv2.hconcat([image, image_2])
    images.append(final_frame)

    # Show image
    plt.figure(figsize=(10, 5))
    plt.imshow(final_frame)
    plt.axis("off")
    plt.show()
    print("\n")

# ==========================================
# STEP 3: Check Image Shape / Dimensions
# ==========================================
print("=== DENSITY & DIMENSIONS ===")
for idx, img in enumerate(images):
    print(
        f"Gambar {idx+1} - Dimensi (Height, Width, Channel): {img.shape}, Tipe Data: {img.dtype}"
    )
print("\n")

# ==========================================
# STEP 4: Modify Pixel Values (Adding Horizontal Line)
# ==========================================
print("=== STEP 4: MANIPULASI PIKSEL CITRA ===")
for idx, img_rgb in enumerate(original_rgb_images):
    modified_img = img_rgb.copy()
    h, w, _ = modified_img.shape

    # Mengubah piksel pada baris tengah menjadi garis horizontal putih [255, 255, 255]
    mid_row = h // 2
    modified_img[mid_row, :] = [255, 255, 255]

    # Concatenate gambar asli dan gambar ter-modifikasi secara horizontal
    comparison_frame = cv2.hconcat([img_rgb, modified_img])

    # Show image
    plt.figure(figsize=(10, 5))
    plt.imshow(comparison_frame)
    plt.title(f"Hasil Step 4 - Gambar {idx+1} (Original vs Modified Pixel)")
    plt.axis("off")
    plt.show()
    print("\n")
```

3. Kegunaan image = cv2.resize(image, (0,0), fx=0.5, fy=0.5) dan pengaruh jika tidak dilakukan
Kegunaan: Memperkecil ukuran (resolusi) gambar menjadi $50\%$ ($0,5$) dari lebar (fx) dan tinggi (fy) aslinya.
Pengaruh jika tidak dilakukan:
Ukuran citra tetap pada resolusi aslinya (bisa terlalu besar saat ditampilkan di layar).
Proses eksekusi/pemrosesan piksel membutuhkan memori dan waktu yang lebih lama.
4. Kegunaan kode [255, 255, 255]
Kode [255, 255, 255] merepresentasikan nilai warna putih pada format RGB/BGR 8-bit.
Masing-masing nilai $255$ menunjukkan intensitas maksimum untuk kanal Red, Green, dan Blue ($R=255, G=255, B=255$).
Pada potongan kode tersebut, nilai [255, 255, 255] digunakan untuk mengubah warna piksel di sepanjang baris tengah menjadi warna putih (membentuk garis horizontal putih).
5. Keterkaitan antara piksel dan resolusi gambar (tinggi/rendah)
Piksel adalah unit atau elemen terkecil pembentuk suatu citra digital.
Resolusi adalah jumlah total piksel yang menyusun citra tersebut ($\text{Lebar} \times \text{Tinggi}$).
Hubungan:
Resolusi Tinggi: Memiliki kerapatan dan jumlah piksel yang sangat banyak. Hasilnya, detail citra terlihat lebih halus, tajam, dan tidak pecah saat diperbesar.
Resolusi Rendah: Memiliki jumlah piksel yang terbatas. Hasilnya, citra cenderung terlihat kabur, kurang detail, dan tampak patah-patah/kotak-kotak (pixelated).


```python
# ==========================================
# STEP 1: Import Library
# ==========================================
import cv2
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from PIL import Image
import skimage
from urllib.request import Request, urlopen

# Setup header request untuk mencegah HTTP Error 521 dari server target
headers = {"User-Agent": "Mozilla/5.0"}


def load_image_from_url(url):
    req_stmt = Request(url, headers=headers)
    req = urlopen(req_stmt)
    arr = np.asarray(bytearray(req.read()), dtype=np.uint8)
    img = cv2.imdecode(arr, cv2.IMREAD_COLOR)
    return img


# ==========================================
# STEP 2: Read & Display Images from URLs
# ==========================================
print("=== STEP 2: BACA & TAMPILKAN CITRA DAFTAR URL ===")
urls = [
    "https://images.unsplash.com/photo-1543466835-00a7907e9de1?w=800",
    "https://images.unsplash.com/photo-1514888286974-6c03e2ca1dba?w=800",
    "https://images.unsplash.com/photo-1533738363-b7f9aef128ce?w=800",
]

images = []
original_rgb_images = []

for url in urls:
    image_bgr = load_image_from_url(url)

    # Resize gambar menjadi setengah ukuran asli
    image_resized = cv2.resize(image_bgr, (0, 0), fx=0.5, fy=0.5)

    # Konversi dari BGR ke RGB
    image_rgb = cv2.cvtColor(image_resized, cv2.COLOR_BGR2RGB)
    original_rgb_images.append(image_rgb.copy())

    # Concatenate horizontal (BGR | RGB)
    final_frame = cv2.hconcat([image_resized, image_rgb])
    images.append(final_frame)

    plt.figure(figsize=(10, 5))
    plt.imshow(final_frame)
    plt.axis("off")
    plt.show()

# ==========================================
# STEP 3: Check Image Shape / Dimensions
# ==========================================
print("\n=== STEP 3: CEK DIMENSI DAN TIPE DATA CITRA ===")
for idx, img in enumerate(images):
    print(
        f"Gambar {idx+1} - Dimensi (Height, Width, Channel): {img.shape}, Tipe Data: {img.dtype}"
    )

# ==========================================
# STEP 4: Modify Pixel Values (Garis Horizontal Putih di Tengah)
# ==========================================
print("\n=== STEP 4: MANIPULASI PIKSEL CITRA (GARIS HORIZONTAL) ===")
sample_img = original_rgb_images[0].copy()
h, w, _ = sample_img.shape

mid_y = h // 2
sample_img[mid_y, :] = [255, 255, 255]

plt.figure(figsize=(10, 5))
plt.imshow(sample_img)
plt.title("Garis Horizontal Putih di Tengah")
plt.axis("off")
plt.show()

# ==========================================
# TUGAS PRAKTIKUM D2
# ==========================================
print("\n=== TUGAS PRAKTIKUM D2 ===")
WHITE = [255, 255, 255]

# --- Tugas 2A: Garis Vertikal di Tengah ---
task2_vert = original_rgb_images[0].copy()
h, w, _ = task2_vert.shape
mid_x = w // 2
task2_vert[:, mid_x] = WHITE

plt.figure(figsize=(10, 5))
plt.imshow(cv2.hconcat([original_rgb_images[0], task2_vert]))
plt.title("Tugas 2A: Garis Vertikal")
plt.axis("off")
plt.show()

# --- Tugas 2B: Garis Menyilang Diagonal ---
task2_diag = original_rgb_images[0].copy()
for i in range(h):
    j1 = int(i * (w - 1) / (h - 1))
    j2 = int((h - 1 - i) * (w - 1) / (h - 1))
    task2_diag[i, j1] = WHITE
    task2_diag[i, j2] = WHITE

plt.figure(figsize=(10, 5))
plt.imshow(cv2.hconcat([original_rgb_images[0], task2_diag]))
plt.title("Tugas 2B: Garis Menyilang Diagonal")
plt.axis("off")
plt.show()

# --- Tugas 3: Garis Horizontal Panjang Tertentu ---
task3_horiz = original_rgb_images[1].copy()
h, w, _ = task3_horiz.shape
mid_y, mid_x = h // 2, w // 2
length = 200
start_x = max(0, mid_x - (length // 2))
end_x = min(w, mid_x + (length // 2))
task3_horiz[mid_y, start_x:end_x] = WHITE

plt.figure(figsize=(10, 5))
plt.imshow(cv2.hconcat([original_rgb_images[1], task3_horiz]))
plt.title("Tugas 3: Garis Horizontal Panjang Tertentu")
plt.axis("off")
plt.show()

# --- Tugas 4: Kotak Piksel Warna Putih ---
task4_box = original_rgb_images[1].copy()
start_y, end_y = h // 3, (h // 3) + 150
start_x, end_x = w // 3, (w // 3) + 150
task4_box[start_y:end_y, start_x:end_x] = WHITE

plt.figure(figsize=(10, 5))
plt.imshow(cv2.hconcat([original_rgb_images[1], task4_box]))
plt.title("Tugas 4: Kotak Piksel Warna Putih")
plt.axis("off")
plt.show()
```
